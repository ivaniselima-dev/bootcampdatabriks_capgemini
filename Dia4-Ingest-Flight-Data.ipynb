# Dia4 - Bronze: Flight Data (COPY INTO)

**Objetivo:** Carregar arquivos JSON de voos da Landing para a tabela Bronze `flight_data` usando `COPY INTO` (carga incremental idempotente).

| Item | Valor |
|---|---|
| **Camada** | Bronze |
| **Entrada** | `/Volumes/{catalog}/landing/files/flight/` (JSON) |
| **Saída** | `{catalog}.bronze.flight_data` |
| **Ordem no pipeline** | Após upload dos JSONs de voos para a Landing |

```python
dbutils.widgets.text("catalog", "capgemini_academy", "Nome do catálogo no Unity Catalog")
catalog = dbutils.widgets.get("catalog")
```

## Carga via COPY INTO

Cria a tabela com `IF NOT EXISTS` e usa `COPY INTO` para carregar os JSONs de voos. O `COPY INTO` rastreia arquivos processados, garantindo idempotência. `mergeSchema = true` permite colunas novas sem quebrar a carga.

```python
%sql
CREATE TABLE IF NOT EXISTS `${catalog}`.`bronze`.`flight_data`;

COPY INTO `${catalog}`.`bronze`.`flight_data`
FROM '/Volumes/${catalog}/landing/files/flight/'
FILEFORMAT = JSON
FORMAT_OPTIONS ('multiline' = 'false')
COPY_OPTIONS ('mergeSchema' = 'true');
```
