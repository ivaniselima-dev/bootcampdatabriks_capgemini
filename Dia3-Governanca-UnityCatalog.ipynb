# Dia3 - Governança: Unity Catalog

**Objetivo:** Criar uma view pública (`vw_movie_ratings_public`) com filmes estatisticamente relevantes e aplicar governança concedendo acesso `SELECT` a um usuário específico.

| Item | Valor |
|---|---|
| **Camada** | Governance (view sobre Silver) |
| **Entrada** | `{catalog}.silver.movie_ratings_summary` |
| **Saída** | `{catalog}.governance.vw_movie_ratings_public` (view) |
| **Ordem no pipeline** | Após `Day2-Movie-Rating-silver` (Silver) |

> **⚠️ Atenção:** A célula de `GRANT` contém o e-mail `contatoivanpe@gmail.com` — ajuste para o usuário correto do seu ambiente antes de executar.

```python
dbutils.widgets.text("catalog", "capgemini_academy", "Nome do catálogo no Unity Catalog")
catalog = dbutils.widgets.get("catalog")
```

## Criação da view governada

Filtra a tabela Silver para manter apenas filmes com ≥10 avaliações, criando uma view pública com `CREATE OR REPLACE VIEW` (idempotente).

```python
%sql
CREATE OR REPLACE VIEW `${catalog}`.`governance`.`vw_movie_ratings_public`
COMMENT 'View pública com filmes de amostra estatisticamente relevante (>=10 avaliações)'
AS
SELECT movieId, title, genres, media_rating, total_avaliacoes
FROM `${catalog}`.`silver`.`movie_ratings_summary`
WHERE total_avaliacoes >= 10;
```

## Concessão de permissão (least privilege)

Concede `SELECT` na view para um usuário específico, simulando o padrão de menor privilégio. Substitua o e-mail pelo usuário correto do seu ambiente.

```python
%sql
-- ⚠️ E-mail pessoal abaixo — troque pelo usuário correto do seu ambiente antes de executar.
GRANT SELECT ON VIEW `${catalog}`.`governance`.`vw_movie_ratings_public` TO `contatoivanpe@gmail.com`;
```
