# Dia3 - Bronze: FIFA World Cup Matches

**Objetivo:** Carregar todos os arquivos CSV de partidas de Copa do Mundo da Landing para a tabela Bronze `fifa_world_cup_all_matches`.

| Item | Valor |
|---|---|
| **Camada** | Bronze |
| **Entrada** | `/Volumes/{catalog}/landing/files/fifa/*.csv` |
| **Saída** | `{catalog}.bronze.fifa_world_cup_all_matches` |
| **Ordem no pipeline** | Após upload dos CSVs para a Landing |

```python
dbutils.widgets.text("catalog", "capgemini_academy", "Nome do catálogo no Unity Catalog")
catalog = dbutils.widgets.get("catalog")
```

## Carga da tabela Bronze

Lê todos os CSVs da pasta FIFA na Landing com `read_files` e inferência de schema, descartando a coluna técnica `_rescued_data`. `CREATE OR REPLACE TABLE` garante idempotência.

```python
%sql
CREATE OR REPLACE TABLE `${catalog}`.`bronze`.`fifa_world_cup_all_matches`
    SELECT * EXCEPT(_rescued_data)
    FROM read_files(
        '/Volumes/${catalog}/landing/files/fifa/*.csv',
        format => 'csv',
        header => true,
        inferSchema => true
    );
```
