# Dia5 - Gold: Top Produtos

**Objetivo:** Agregar vendas por produto a partir da Silver `pedidos_autoloader_v2` (versão deduplicada), calculando quantidade vendida e receita por produto.

| Item | Valor |
|---|---|
| **Camada** | Gold |
| **Entrada** | `{catalog}.silver.pedidos_autoloader_v2` |
| **Saída** | `{catalog}.gold.top_produtos` |
| **Ordem no pipeline** | Após `Dia5-Pedidos-Bronze-Silver` (Silver v2) |

```python
dbutils.widgets.text("catalog", "capgemini_academy", "Nome do catálogo no Unity Catalog")
catalog = dbutils.widgets.get("catalog")
```

## Agregação Gold — Top produtos

Agrupa vendas por `produto_id` a partir da Silver v2 (deduplicada). `CREATE OR REPLACE TABLE` garante idempotência.

```python
%sql
CREATE OR REPLACE TABLE `${catalog}`.`gold`.`top_produtos` AS
    SELECT
        produto_id,
        SUM(quantidade) AS qtde_vendida,
        SUM(valor_total) AS receita
    FROM `${catalog}`.`silver`.`pedidos_autoloader_v2`
    GROUP BY produto_id;
```
